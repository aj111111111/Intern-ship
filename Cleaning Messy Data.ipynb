# Clean a messy dataset in Pandas
handling missing values, filtering rows, creating new columns


```python
import pandas as pd
```

```python
df = pd.read_csv(r"data.csv")
df
```

#### 1) Checking Duplicate rows 

```python
df[df.duplicated(keep=False)]
```

```python
# Drop duplicated rows
df = df.drop_duplicates()
df
```

```python
df['Date']
```

#### 2) Removing (') 
at starting & at end if there

```python
df["Date"] = df["Date"].str.strip("'")
df["Date"]
```

```python
df
```

#### 3) Checking total null in all column

```python
df.isnull().sum() 
```

#### 3) Replacing empty cell first with naN.
Then output all null rows.

```python
import numpy as np

null_rows = df.replace(r'^\s*$', np.nan, regex=True)
null_rows = null_rows[null_rows.isnull().any(axis=1)]
null_rows
```

#### 4) Dropping NaN or Empty row

```python
import numpy as np
# Replace empty/whitespace-only strings with NaN, then drop all null rows
df = df.replace(r'^\s*$', np.nan, regex=True).dropna()
df
```

#### 5) Resetting Index

```python
df.reset_index(inplace= True)
df.head()
```

```python
df.columns
```

#### 6) Keeping important columns

```python
df2 = df[['Duration', 'Date', 'Pulse', 'Maxpulse', 'Calories']]
df2.head()
```

#### 7) Using a regex pattern 
to find any 8-digit date string (like 20201201), format it with slashes (2020/12/01), and leave all other values completely untouched.

```python
df2['Date'] = df2['Date'].replace(r'^(\d{4})(\d{2})(\d{2})$', r'\1/\2/\3', regex=True)
df2
```

```python
df2['Duration'] = df2['Duration'].astype(str).str.replace(r'(\d{2})\d+', r'\1', regex=True)
df2

# pattern r'(\d{2})\d+' targets numbers with 3 or more digits and keep only first two digit digit
# Earlier at row 7 (Duration column) had 450 invalid value.
```

#### 8) Saving Cleaned data to "Cleaned-data.xlsx" excel with sheet name - "Cleaned Data"

```python
import os

excel_filename = "Cleaned-Data.xlsx"

if os.path.exists(excel_filename):
    with pd.ExcelWriter(excel_filename, engine='openpyxl', mode='a', if_sheet_exists='replace') as writer:
        df2.to_excel(writer, index=False, sheet_name="Cleaned Data")
else:
    df2.to_excel(excel_filename, index=False, sheet_name="Cleaned Data")

print(f"Data successfully saved to {excel_filename}!")
```

```python
df2.to_csv('Cleaned Data.csv', index=False)
```
