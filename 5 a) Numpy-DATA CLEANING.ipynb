## Missing Data  - empty value, or a number 0, or an invalid value (a string for example) can be considered "missing data". 

```python
import numpy as np
import pandas as pd
```

```python
falsy_values = (0, False, None, '', [], {})
```

For Python, all the values above are considered "falsy":

```python
any(falsy_values)   # "any" function return true if any value is true in an iterable (like a list, tuple, or set)
```

```python
np.nan    #Numpy has a special "nullable" value for numbers which is np.nan. It's NaN: "Not a number"
```

 The np.nan value is kind of a virus. Everything that it touches becomes np.nan:

```python
print(3 + np.nan)

a = np.array([1, 2, 3, np.nan, np.nan, 4])
print(a.sum())

print(a.mean())
```

```python
print(3 + None)
```

```python
a = np.array([1, 2, 3, np.nan, None, 4], dtype='float')
print(a)

# As we said, np.nan is like a virus. If you have any nan value in an array and you try to
#perform an operation on it, you'll get unexpected results: 

a = np.array([1, 2, 3, np.nan, np.nan, 4])


print(a.mean())
print(a.sum())

# Numpy also supports an "Infinite" type:
# np.inf
# Which also behaves as a virus:

print(3 + np.inf)
print(np.inf / 3)
print(np.inf / np.inf)
print(3 + np.nan)
```

```python
b = np.array([1, 2, 3, np.inf, np.nan, 4], dtype=float)
print(b.sum())
```

```python
### It means numpy has "nan" & "inf" to display "Not a number" & "Infinite" resp.
```

## Checking for nan or inf

```python
np.isnan(np.nan)
```

```python
np.isinf(np.inf)
```

And the joint operation can be performed with np.isfinite.

```python
np.isfinite(np.nan), np.isfinite(np.inf)
```

np.isnan and np.isinf also take arrays as inputs, and return boolean arrays as results:

```python
np.isnan(np.array([1, 2, 3, np.nan, np.inf, 4]))
```

```python
np.isinf(np.array([1, 2, 3, np.nan, np.inf, 4]))
```

```python
np.isfinite(np.array([1, 2, 3, np.nan, np.inf, 4]))
```

# Filtering out 

```python
a = np.array([1, 2, 3, np.nan, np.nan, 4])
```

```python
np.isnan(a)
```

```python
a[~np.isnan(a)]                 # Filtering out only not Nan
```

```python
a[np.isfinite(a)].sum()         # summing only which is finite numbers
```

```python
a[np.isfinite(a)].mean()
```
