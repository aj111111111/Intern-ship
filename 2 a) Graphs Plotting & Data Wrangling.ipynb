<h1><b>5) Column wrangling </b></h1>
We can also create new columns or modify existing ones.

### Add and calculate a new `Revenue_per_Age` column

```python
sales['Revenue_per_Age'] = sales['Revenue'] / sales['Customer_Age']

sales['Revenue_per_Age'].head()
```

```python
sales['Revenue_per_Age'].plot(kind='density', figsize=(14,6))
```

### Add and calculate a new `Calculated_Cost` column

Use this formula

$$ Calculated\_Cost = Order\_Quantity * Unit\_Cost $$

```python
sales['Calculated_Cost'] = sales['Order_Quantity'] * sales['Unit_Cost']

sales['Calculated_Cost'].head()
```

```python
(sales['Calculated_Cost'] != sales['Cost']).sum()
```

```python
sales.head()
```

```python
sales.plot(kind='scatter', x='Calculated_Cost', y='Profit', figsize=(6,6))
```

<h1><b>6) Add and calculate a new `Calculated_Revenue` column </b></h1>


Use this formula

$$ Calculated\_Revenue = Cost + Profit $$

```python
sales['Calculated_Revenue'] = sales['Cost'] + sales['Profit']

sales['Calculated_Revenue'].head()
```

```python
(sales['Calculated_Revenue'] != sales['Revenue']).sum()
```

sales['Calculated_Revenue'] != sales['Revenue']: Compares the two columns row by row, returning a boolean Series of True (where the values differ) and False (where they match).

.sum(): Treats True as 1 and False as 0, summing them up to give the total count of discrepancies between the two columns.

### How many records belong to Age Group `Youth (<25)` or `Adults (35-64)`?

```python
sales.loc[(sales['Age_Group'] == 'Youth (<25)') | (sales['Age_Group'] == 'Adults (35-64)')].shape
```

```python
sales.loc[(sales['Age_Group'] == 'Youth (<25)') | (sales['Age_Group'] == 'Adults (35-64)')].shape[0]   #above is to see, this is right
```

It returns the total count of rows that match your filtering conditions (i.e., the number of records where the age group is either 'Youth (<25)' or 'Adults (35-64)').

sales.loc[...]: Filters the DataFrame to include only the rows where the specified conditions are met.

.shape: Returns a tuple representing the dimensions of the filtered DataFrame in the format (number_of_rows, number_of_columns).

[0]: Extracts the first element of that tuple, isolating just the number of rows.
