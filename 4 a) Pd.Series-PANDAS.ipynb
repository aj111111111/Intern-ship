# Main

```python
pd.Series({
    'Canada': 35.467,
    'France': 63.951,
    'Germany': 80.94,
    'Italy': 60.665,
    'Japan': 127.061,
    'United Kingdom': 64.511,
    'United States': 318.523
}, name='Population in millions')
```

```python
original_by_countryname = pd.Series(
    [35.467, 63.951, 80.94, 60.665, 127.061, 64.511, 318.523],
    index=['Canada', 'France', 'Germany', 'Italy', 'Japan', 'United Kingdom',
       'United States'],
    name='Population in millions')
```

# Conditional selection (boolean arrays)

```python
population
```

```python
population > 64                     # Boolean Condition
```

```python
population[population > 64]           # Getting by condition, a filtering
```

```python
population.mean()
```

```python
population[population > population.mean()]        #  filtering
```

```python
population.std()
```

#####  ~ not  ------  | or  ---------   & and

```python
population[(population > population.mean() - population.std() / 2) | (population > population.mean() + population.std() / 2)]
```
